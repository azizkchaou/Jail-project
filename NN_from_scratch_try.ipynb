```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder
from scipy.sparse import hstack

# Re-define df to ensure it's available in case of kernel restart or out-of-order execution
df = pd.read_csv("/kaggle/input/datasets/alikchaou/sent-filtered/sent_filtered.csv")
df = df[df['LAW_ENFORCEMENT_AGENCY'].isna()==False]

# Define categorical columns to be one-hot encoded
ohe_cols = ['DISPOSITION_CHARGED_OFFENSE_TITLE', 'LAW_ENFORCEMENT_AGENCY']

# Define numerical/pre-encoded columns
numeric_cols = [
    'CHARGE_COUNT',
    'GENDER_Female',
    'GENDER_Male',
    'RACE_ASIAN',
    'RACE_BLACK',
    'RACE_HISPANIC',
    'RACE_WHITE',
    'RACE_WHITE [HISPANIC OR LATINO]',
    'RACE_WHITE/BLACK [HISPANIC OR LATINO]',
    'CHARGE_DISPOSITION_Finding Guilty',
    'CHARGE_DISPOSITION_Plea Of Guilty',
    'CHARGE_DISPOSITION_Verdict Guilty'
]

# Initialize OneHotEncoder with sparse_output=True to save memory
encoder = OneHotEncoder(sparse_output=True, handle_unknown='ignore')

# Fit and transform the categorical columns
encoded_features_sparse = encoder.fit_transform(df[ohe_cols])

# Extract the numerical columns and convert them to float type
numeric_features_dense = df[numeric_cols].astype(float).values

# Combine the sparse encoded features with the dense numerical features using scipy.sparse.hstack
# This will result in a sparse matrix
X1 = hstack([encoded_features_sparse, numeric_features_dense])
```

```python
def init_parameters(layer_dims):
    params = {}
    L = len(layer_dims)
    for l in range(1, L):
        n_in  = layer_dims[l - 1]
        n_out = layer_dims[l]
        is_output = (l == L - 1)
        scale = np.sqrt(1 / n_in) if is_output else np.sqrt(2 / n_in)
        params[f'W{l}'] = np.random.randn(n_out, n_in) * scale
        params[f'b{l}'] = np.zeros((n_out, 1))

        # Add Batch Norm parameters for hidden layers
        if not is_output:
            params[f'gamma{l}'] = np.ones((n_out, 1))
            params[f'beta{l}'] = np.zeros((n_out, 1))
            # Running averages for inference
            params[f'running_mean{l}'] = np.zeros((n_out, 1))
            params[f'running_var{l}'] = np.ones((n_out, 1))
    return params
```

```python
def batchnorm_forward(Z, gamma, beta, running_mean, running_var, training, epsilon=1e-8, momentum=0.9):
    if training:
        mu = np.mean(Z, axis=1, keepdims=True)
        var = np.var(Z, axis=1, keepdims=True)

        Z_norm = (Z - mu) / np.sqrt(var + epsilon)
        out = gamma * Z_norm + beta

        # Update running stats
        running_mean[:] = momentum * running_mean + (1 - momentum) * mu
        running_var[:] = momentum * running_var + (1 - momentum) * var

        cache = (Z, Z_norm, mu, var, gamma, beta, epsilon)
    else:
        Z_norm = (Z - running_mean) / np.sqrt(running_var + epsilon)
        out = gamma * Z_norm + beta
        cache = None

    return out, cache

def linear_forward(A, W, b):
    Z = np.dot(W, A) + b
    return Z, (A, W, b)

def softmax(z):
    z_shifted = z - np.max(z, axis=0, keepdims=True)
    exp_z = np.exp(z_shifted)
    return exp_z / np.sum(exp_z, axis=0, keepdims=True), z_shifted

def leaky_relu(Z, alpha=0.01):
    return np.where(Z > 0, Z, alpha * Z), Z

def L_model_forward(X, parameters, alpha, keep_prob=0.8, training=True):
    caches = []
    dropout_masks = []
    A = X
    L = len(parameters) // 6 # W, b, gamma, beta, running_mean, running_var (roughly)
    # Correct L calculation based on parameter keys
    L = max([int(k[1:]) for k in parameters.keys() if k.startswith('W')])

    for l in range(1, L):
        A_prev = A
        Z, lin_cache = linear_forward(A_prev, parameters[f'W{l}'], parameters[f'b{l}'])

        # Batch Norm
        Z_bn, bn_cache = batchnorm_forward(Z, parameters[f'gamma{l}'], parameters[f'beta{l}'],
                                           parameters[f'running_mean{l}'], parameters[f'running_var{l}'], training)

        A, act_cache = leaky_relu(Z_bn, alpha)
        caches.append((lin_cache, bn_cache, act_cache))

        if training:
            A, mask = dropout_forward(A, keep_prob)
            dropout_masks.append(mask)
        else:
            dropout_masks.append(None)

    # Output Layer (No BN or Dropout)
    ZL, lin_cache = linear_forward(A, parameters[f'W{L}'], parameters[f'b{L}'])
    AL, act_cache = softmax(ZL)
    caches.append((lin_cache, None, act_cache))

    return AL, caches, dropout_masks
```

```python
def compute_focal_loss(AL, Y, gamma=2.0):
    m = Y.shape[1]
    eps = 1e-15
    pt = np.sum(Y * AL, axis=0, keepdims=True)
    focal_weight = (1 - pt) ** gamma
    cost = - (1/m) * np.sum(focal_weight * Y * np.log(AL + eps))
    return np.squeeze(cost)
```

```python
def batchnorm_backward(dout, cache):
    Z, Z_norm, mu, var, gamma, beta, eps = cache
    m = Z.shape[1]

    dgamma = np.sum(dout * Z_norm, axis=1, keepdims=True)
    dbeta = np.sum(dout, axis=1, keepdims=True)

    dZ_norm = dout * gamma
    dvar = np.sum(dZ_norm * (Z - mu) * -0.5 * (var + eps)**(-1.5), axis=1, keepdims=True)
    dmu = np.sum(dZ_norm * -1/np.sqrt(var + eps), axis=1, keepdims=True) + dvar * np.mean(-2 * (Z - mu), axis=1, keepdims=True)

    dZ = (dZ_norm / np.sqrt(var + eps)) + (dvar * 2 * (Z - mu) / m) + (dmu / m)
    return dZ, dgamma, dbeta

def linear_backward(dZ, cache):
    A_prev, W, b = cache
    m = A_prev.shape[1]
    dW = (1/m) * np.dot(dZ, A_prev.T)
    db = (1/m) * np.sum(dZ, axis=1, keepdims=True)
    dA_prev = np.dot(W.T, dZ)
    return dA_prev, dW, db

def L_model_backward(AL, Y, caches, alpha, dropout_masks, keep_prob=0.8):
    grads = {}
    L = len(caches)
    m = AL.shape[1]

    # Focal loss gradient
    gamma_focal = 2.0
    pt = np.sum(Y * AL, axis=0, keepdims=True)
    focal_weight = (1 - pt) ** gamma_focal
    dZ = focal_weight * (AL - Y)

    # Output layer backward
    lin_cache, _, _ = caches[L-1]
    grads[f"dA{L-1}"], grads[f"dW{L}"], grads[f"db{L}"] = linear_backward(dZ, lin_cache)

    for l in reversed(range(L-1)):
        lin_cache, bn_cache, act_cache = caches[l]
        dA = grads[f"dA{l+1}"]

        if dropout_masks[l] is not None:
            dA = dropout_backward(dA, dropout_masks[l], keep_prob)

        dZ_bn = np.where(act_cache > 0, dA, alpha * dA)
        dZ, dgamma, dbeta = batchnorm_backward(dZ_bn, bn_cache)

        dA_prev, dW, db = linear_backward(dZ, lin_cache)
        grads[f"dA{l}"] = dA_prev
        grads[f"dW{l+1}"] = dW
        grads[f"db{l+1}"] = db
        grads[f"dgamma{l+1}"] = dgamma
        grads[f"dbeta{l+1}"] = dbeta

    return grads
```

```python
def init_adam(parameters):
    v, s = {}, {}
    for key, val in parameters.items():
        if 'running' not in key:
            v[f'd{key}'] = np.zeros_like(val)
            s[f'd{key}'] = np.zeros_like(val)
    return v, s

def update_parameters_adam(params, grads, v, s, t, lr=0.001, b1=0.9, b2=0.999, eps=1e-8):
    for key in params.keys():
        if 'running' in key: continue

        g = grads[f'd{key}']
        v[f'd{key}'] = b1 * v[f'd{key}'] + (1-b1) * g
        s[f'd{key}'] = b2 * s[f'd{key}'] + (1-b2) * g**2
        v_corr = v[f'd{key}'] / (1 - b1**t)
        s_corr = s[f'd{key}'] / (1 - b2**t)
        params[key] -= lr * v_corr / (np.sqrt(s_corr) + eps)
    return params, v, s
```

```python
def dropout_forward(A, keep_prob=0.8):
    mask = (np.random.rand(*A.shape) < keep_prob)
    A_dropped = (A * mask) / keep_prob
    return A_dropped, mask

def dropout_backward(dA, mask, keep_prob=0.8):
    return (dA * mask) / keep_prob
```

```python
from sklearn.metrics import classification_report, f1_score
input_features = X_train.shape[1]
output_classes = len(np.unique(Yc_train))


layer_dims = [
    input_features,  # 2271
    512,
    256,
    64,
    output_classes   # 7
]
parameters = init_parameters(layer_dims)

def training(X_train_sparse, Yc_train_labels, parameters, learning_rate,  num_iterations, alpha, output_classes, X_test_t, Yc_test, batch_size=256):
    costs = []
    m = Yc_train_labels.shape[0]

    # Correctly create one-hot labels for the entire training set
    Y_one_hot_full = np.eye(output_classes)[Yc_train_labels].T

    v, s = init_adam(parameters)
    t = 0

    for i in range(num_iterations):
        epoch_cost = 0
        num_batches = 0

        # Shuffle indices for each epoch
        permutation = np.random.permutation(m)
        X_shuffled = X_train_sparse[permutation, :]
        Y_shuffled = Y_one_hot_full[:, permutation]
        Y_labels_shuffled = Yc_train_labels[permutation]

        for j in range(0, m, batch_size):
            X_batch_sparse = X_shuffled[j:j+batch_size, :]
            Y_batch = Y_shuffled[:, j:j+batch_size]
            X_batch_dense = X_batch_sparse.toarray().T

            # Forward propagation
            AL, caches, dropout_masks = L_model_forward(X_batch_dense, parameters, alpha, keep_prob=0.8, training=True)

            # Cost calculation
            batch_cost = compute_focal_loss(AL, Y_batch, gamma=2.0)
            epoch_cost += batch_cost
            num_batches += 1

            # Backward propagation
            grads = L_model_backward(AL, Y_batch, caches, alpha, dropout_masks, keep_prob=0.8)

            t += 1
            parameters, v, s = update_parameters_adam(parameters, grads, v, s, t, lr=learning_rate)

        avg_epoch_cost = epoch_cost / num_batches
        costs.append(avg_epoch_cost)

        if i % 10 == 0 or i == num_iterations - 1:
            print(f"\n--- Epoch {i} ---")
            print(f"Cost: {avg_epoch_cost:.6f}")

            # Training F1 (on the last batch)
            train_preds = np.argmax(AL, axis=0)
            train_true = Y_labels_shuffled[j:j+batch_size]
            train_f1 = f1_score(train_true, train_preds, average='weighted', zero_division=0)
            print(f"Training Weighted F1 (Last Batch): {train_f1:.4f}")

            # Test Evaluation
            AL_test, _, _ = L_model_forward(X_test_t, parameters, alpha, keep_prob=0.8, training=False)
            test_preds = np.argmax(AL_test, axis=0)
            print("Test Classification Report:")
            print(classification_report(Yc_test, test_preds, zero_division=0))

    return parameters, costs

# Execution
initial_learning_rate = 0.001
num_iterations = 120
alpha = 0.01
batch_size = 2048

scaler = MaxAbsScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
X_test_t = X_test_scaled.toarray().T
print("Input features:", input_features)
print("Train shape:", X_train.shape)

print("Starting neural network training...")
trained_parameters, costs = training(X_train_scaled, Yc_train, parameters, initial_learning_rate, num_iterations, alpha, output_classes, X_test_t, Yc_test, batch_size=batch_size)
print("Training finished.")
```
